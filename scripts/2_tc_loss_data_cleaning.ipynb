**Join State and US Tree Loss Datasets**



```r
#read in Global Nature Watch tree cover loss by state spreadsheet
df_states <- read_excel("/content/drive/MyDrive/data_projects/tree_birds/raw_data/USA.xlsx", sheet = "Subnational 1 tree cover loss")

#preview data
head(df_states)
```

```r
#combine yearly US tree cover loss data with state data

#load total US data
df_us <- read_excel("/content/drive/MyDrive/data_projects/tree_birds/raw_data/USA.xlsx", sheet = "Country tree cover loss") %>%
#rename df_us$country column to subnational1 to enable binding datasets
  rename(subnational1 = country)

#remove country column from state data to enable binding datasets
df_states$country <- NULL

#bind US data with state data
df_united_states <- rbind(df_states, df_us)
```

**Process Tree Cover Loss Data**



```r
#filter data by 30 percent threshold tree canopy density
df_states_30 <- df_united_states %>%
  filter(threshold == "30")

#remove threshold column
df_states_30$threshold <- NULL
#remove area_ha column
df_states_30$area_ha <- NULL
#remove 2010 extent column
df_states_30$extent_2010_ha <- NULL
```

```r
#pivot data into one year column

#remove text from years
df_states_long <- df_states_30 %>%
  rename(
    `2001` = tc_loss_ha_2001,
    `2002` = tc_loss_ha_2002,
    `2003` = tc_loss_ha_2003,
    `2004` = tc_loss_ha_2004,
    `2005` = tc_loss_ha_2005,
    `2006` = tc_loss_ha_2006,
    `2007` = tc_loss_ha_2007,
    `2008` = tc_loss_ha_2008,
    `2009` = tc_loss_ha_2009,
    `2010` = tc_loss_ha_2010,
    `2011` = tc_loss_ha_2011,
    `2012` = tc_loss_ha_2012,
    `2013` = tc_loss_ha_2013,
    `2014` = tc_loss_ha_2014,
    `2015` = tc_loss_ha_2015,
    `2016` = tc_loss_ha_2016,
    `2017` = tc_loss_ha_2017,
    `2018` = tc_loss_ha_2018,
    `2019` = tc_loss_ha_2019,
    `2020` = tc_loss_ha_2020,
    `2021` = tc_loss_ha_2021,
    `2022` = tc_loss_ha_2022,
    `2023` = tc_loss_ha_2023,
    `2024` = tc_loss_ha_2024,
    `2025` = tc_loss_ha_2025) %>%
  #pivot by year
  pivot_longer(
    cols = `2001`:`2025`,
    names_to = "year",
    values_to = "yearly_loss_ha"
  )

  #covert year column into integers to enable joining datasets
 df_states_long$year <- as.integer(df_states_long$year)

glimpse(df_states_long)
```

**Combine US and State Driver (Reason for Tree Loss) Data**



```r
#load Global Nature Watch tree cover loss state drivers spreadsheet
df_drivers <- read_excel("/content/drive/MyDrive/data_projects/tree_birds/raw_data/USA.xlsx", sheet = "Subnational 1 drivers")
```

```r
#combine US driver data with state driver data

#load total US drivers data
us_drivers <- read_excel("/content/drive/MyDrive/data_projects/tree_birds/raw_data/USA.xlsx", sheet = "Country drivers") %>%
#rename df_us$country column to subnational1 to enable binding datasets
  rename(subnational1 = country)

#remove country column from state data to enable binding datasets
df_drivers$country <- NULL

#bind US data with state data
united_drivers <- rbind(df_drivers, us_drivers)
```

```r
#preview data
head(united_drivers)
```

**Preprocess Tree Cover Loss Driver Data**

```r
#split and sum tc loss driver values by loss type

#pivot wide
united_wide <- united_drivers %>%
  pivot_wider(
    names_from = driver,
    values_from = tc_loss_ha
  ) %>%

#add tc loss by temporary and permanent causes
mutate(
  temp_tc_loss_ha = Logging + Wildfire + `Other natural disturbances`,
  perm_tc_loss_ha = `Permanent agriculture` + `Settlements & Infrastructure` + `Hard commodities`) %>%

#move columns for easier visual grouping
relocate(`Hard commodities`, .before = Wildfire) %>%
  relocate(Wildfire, .after = Logging) %>%
  relocate(temp_tc_loss_ha, .before = Logging) %>%
  relocate(perm_tc_loss_ha, .before = `Permanent agriculture`)
```

```r
#turn NAs into 0s to account for states that didn't report on or lose tree cover from every type of loss
united_wide[is.na(united_wide)] <- 0
```

```r
#remove threshold and country column
united_wide$threshold <- NULL
united_wide$country <- NULL

#move year column left
united_drivers_organized <- united_wide %>% relocate(year, .before = temp_tc_loss_ha)
```

**Combine State Tree Loss Data with Driver (Reason for Loss) Data**

```r
#join tree loss data to drivers data by year
df_forestloss <- left_join(x = df_states_long, y = united_drivers_organized, by = c("subnational1", "year"))

head(df_forestloss)
```

```r
#standardize column names to lowercase and rename for brevity and understandability
df_forest_lc <- df_forestloss %>%
  rename(common_region = subnational1,
        tc_2000 = extent_2000_ha,
        logging = Logging,
        wildfire = Wildfire,
        other_nature = `Other natural disturbances`,
        development = `Settlements & Infrastructure`,
        perm_ag = `Permanent agriculture`,
        hard_commod = `Hard commodities`)
```

```r
#create cumulative loss columns for graphing
df_trees <- df_forest_lc%>%
  group_by(common_region) %>%
  mutate(cum_yearly_loss_ha = cumsum(yearly_loss_ha)) %>%
  mutate(cum_temp_loss_ha = cumsum(temp_tc_loss_ha)) %>%
  mutate(cum_perm_loss_ha = cumsum(perm_tc_loss_ha)) %>%
  ungroup()
```

```r
#review changes
glimpse(df_trees)
```

```r
#write to clean data folder
write_csv(df_trees, "clean_data/tree_loss_data_2001-2025.csv")

# Verify the file exists
list.files("clean_data/")
```

**Combine Tree Cover Data with BBS Dataset**

```r
#read in bbs_data from 1_bbs_data_cleaning
bbs_data <- read.csv("clean_data/north_american_bird_abundance_1966-2025_clean.csv")
```

```r
#join the bbs data set with the tree cover loss data set to make a complete data set
df_complete <- left_join(x = bbs_data, y = df_trees, by = c("common_region", "year"))
```

```r
#remove any duplicated rows
df_states <- df_complete[!duplicated(df_complete), ] %>%

#filter out all years before 2001, regions missing forest loss data, and any other NAs
filter(!is.na(yearly_loss_ha), !is.na(perm_tc_loss_ha), !is.na(temp_tc_loss_ha))

cat("Total rows of data:", nrow(df_complete), "\n")
cat("Rows after cleaning:", nrow(df_states), "\n")
cat("Rows dropped:", nrow(df_complete) - nrow(df_states), "\n")
```
