**Add Common Names for Taxonomic Order and Family**

```r
#simplify common_name and remove seq and french name columns from bbs_species_list
bbs_species_english <- full_species_list %>%
  rename(common_name = english_common_name) %>%
  select(aou, common_name, order, family, genus, species)
```

```r
#Create dataframe of common names for taxonomnic/scientific bird orders
common_order_list <- data.frame(
  order = c("Accipitriformes", "Anseriformes", "Apodiformes", "Caprimulgiformes", "Cathartiformes", "Charadriiformes", "Ciconiiformes", "Columbiformes", "Coraciiformes", "Cuculiformes", "Falconiformes", "Galliformes", "Gaviiformes", "Gruiformes", "Passeriformes", "Pelecaniformes", "Piciformes", "Podicipediformes", "Procellariiformes", "Psittaciformes", "Strigiformes", "Suliformes", "Trogoniformes"),
  common_order = c("Hawks, Eagles, Kites, and Allies", "Waterfowl", "Hummingbirds, Swifts, and Treeswifts", "Nightjars and Nighthawks", "New World Vultures", "Shorebirds and Allies", "Storks", "Pigeons and Doves", "Kingfishers and Allies", "Cuckoos and Allies", "Falcons", "Landfowl", "Loons", "Cranes, Rails, and Allies", "Perching Birds/Songbirds", "Pelicans, Herons, Ibises, and Allies", "Woodpeckers and Allies", "Grebes", "Tubenoses", "Parrots", "Owls", "Frigates, Cormorants, and Allies", "Trogons and Quetzals")
)

head(common_order_list)
```

```r
#load .csv of common names for taxonomic/scientific bird families
common_family_list <- read.csv("/content/drive/MyDrive/data_projects/tree_birds/raw_data/bird_families.csv")

#preview data
head(common_family_list)
```

```r
#Join common names for taxonomic bird orders and families to bbs_species_english
bbs_data_order <- left_join(x = bbs_species_english, y = common_order_list, by = "order")

amended_species_list <- left_join(x = bbs_data_order, y = common_family_list, by = "family")
```

**Join Amended Species List and Common Region Data to BBS Survey Index**

```r
#read in the USGS Breeding Bird Survey incides (relative population abundance) 1966-2025 csv
bbs_indices <- read.csv("/content/drive/MyDrive/data_projects/tree_birds/raw_data/bbs_data/2026_dataset/2026_analysis/BBS_Core_Indices_1966-2025.csv")

#preview bbs_indices data
head(bbs_indices)
```

```r
#standardize bbs column names to lowercase, change index to relative_abundance to be more descriptive
bbs_renamed <- bbs_indices %>% rename(
  aou = AOU,
  region = Region,
  year = Year,
  relative_abundance = Index,
  x2.5.ci = X2.5.CI,
  x97.5.ci = X97.5.CI
)
```

```r
#join bbs_indices with amended species data by AOU column
bbs_data_species <- left_join(x = amended_species_list, y = bbs_renamed, by = "aou")

#remove AOU column
bbs_data_species$aou <- NULL
```

```r
#read in bbs regions data
bbs_regions <- read.csv("/content/drive/MyDrive/data_projects/tree_birds/raw_data/bbs_regions.csv")

#preview data
head(bbs_regions)
```

```r
#combine bbs data with common names for Bird Conservation Regions and states
bbs_data_joined <- left_join(x = bbs_data_species, y = bbs_regions, by = "region") %>%
  #move Common_Region column after Region column for easier use
  relocate(common_region, .after = region)
```

**Clean Data**

```r
#remove any duplicated rows
bbs_data_single <- bbs_data_joined[!duplicated(bbs_data_joined), ]

#remove missing data
bbs_data_clean <- bbs_data_single %>%
  filter(!is.na(relative_abundance), !is.na(common_name))

#determine number of columns cleaned
cat("Rows in raw data:", nrow(bbs_data_joined), "\n")
cat("Rows after cleaning:", nrow(bbs_data_clean), "\n")
cat("Rows dropped:", nrow(bbs_data_joined) - nrow(bbs_data_clean), "\n")
```

```r
#review changes to bbs data
glimpse(bbs_data_clean)
```

```r
#write combined bird information csv to clean drive
write_csv(bbs_data_clean, "clean_data/north_american_bird_abundance_1966-2025_clean.csv")
```
